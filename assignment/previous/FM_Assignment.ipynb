```python
import os
from pathlib import Path

train_raw = Path("/content/drive/My Drive/FoundationModel-Assignment/UW-DiffPhys/data/EUVP/train/raw")
train_ref = Path("/content/drive/My Drive/FoundationModel-Assignment/UW-DiffPhys/data/EUVP/train/ref")

raw_files = sorted([p.name for p in train_raw.iterdir() if p.is_file()])
ref_files = sorted([p.name for p in train_ref.iterdir() if p.is_file()])

print("RAW:", len(raw_files))
print("REF:", len(ref_files))

print("\nFirst 10 RAW:")
print(raw_files[:10])

print("\nFirst 10 REF:")
print(ref_files[:10])

#%cd /content/UW-DiffPhys
os.chdir("/content/drive/My Drive/FoundationModel-Assignment/UW-DiffPhys")

from types import SimpleNamespace
import yaml
import datasets

with open("configs/underwater_euvp_128.yml", "r") as f:
    cfg_dict = yaml.safe_load(f)


def dict_to_namespace(d):
    if isinstance(d, dict):
        return SimpleNamespace(
            **{k: dict_to_namespace(v) for k, v in d.items()}
        )
    return d


config = dict_to_namespace(cfg_dict)

dataset = datasets.Underwater(config)

train_loader, val_loader = dataset.get_loaders()

print("Train batches:", len(train_loader))
print("Val batches:", len(val_loader))
```
